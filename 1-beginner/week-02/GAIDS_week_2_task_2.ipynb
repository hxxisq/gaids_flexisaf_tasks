### Exercise 1: Read Total profit of all months and show it using a line plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('company_sales_data.csv')
```

```python
df['total_profit']
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df['month_number'],df['total_profit'],
         color='blue', marker='o', linestyle='-', linewidth=2, markersize=6)
plt.title('Total profit of all months')
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.xticks(df['month_number'])
plt.tight_layout()
plt.show()
```

### Exercise 2: Read data for Bathing soap and facewash of all months and display it using the Subplot.

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))

axes[0].plot(df['month_number'], df['bathingsoap'],
             color='green', marker='o', linestyle='-', linewidth=2)
axes[0].set_title("Bathing Soap Sales Per Month")
axes[0].set_xlabel("Month Number")
axes[0].set_ylabel("Units Sold")
axes[0].set_xticks(df['month_number'])

axes[1].plot(df['month_number'], df['facewash'],
             color='green', marker='o', linestyle='-', linewidth=2)
axes[1].set_title("Facewash Sales Per Month")
axes[1].set_xlabel("Month Number")
axes[1].set_ylabel("Units Sold")
axes[1].set_xticks(df['month_number'])

plt.tight_layout()
plt.show()
```

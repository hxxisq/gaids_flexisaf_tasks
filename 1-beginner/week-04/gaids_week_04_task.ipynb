```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the built-in Flights dataset
flights = sns.load_dataset('flights')
flights.head()
```

```python
# Prepare data for the line plot (calculating total passengers per year)
yearly_flights = flights.groupby('year')['passengers'].sum().reset_index()
yearly_flights.head()
```

```python
# Set up a 2x2 grid for the visual layout
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Airline Passengers Data Visualization', fontsize=16, fontweight='bold')

# --- TOP LEFT: Line Plot (Matplotlib) ---
axes[0, 0].plot(yearly_flights['year'], yearly_flights['passengers'], marker='o', color='dodgerblue', linewidth=2)
axes[0, 0].set_title('Total Passengers per Year (Line Plot)')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Total Passengers')
axes[0, 0].grid(True, linestyle='--', alpha=0.6)

# --- TOP RIGHT: Scatter Plot (Matplotlib) ---
axes[0, 1].scatter(flights['year'], flights['passengers'], color='tomato', alpha=0.7)
axes[0, 1].set_title('Monthly Passengers Over Time (Scatter Plot)')
axes[0, 1].set_xlabel('Year')
axes[0, 1].set_ylabel('Passengers per Month')
axes[0, 1].grid(True, linestyle='--', alpha=0.6)

# --- BOTTOM LEFT: Histogram (Seaborn) ---
sns.histplot(data=flights, x='passengers', kde=True, ax=axes[1, 0], color='mediumpurple')
axes[1, 0].set_title('Distribution of Passenger Volume (Histogram)')
axes[1, 0].set_xlabel('Number of Passengers')
axes[1, 0].set_ylabel('Frequency')

# --- BOTTOM RIGHT: Box Plot (Seaborn) ---
sns.boxplot(data=flights, x='month', y='passengers', ax=axes[1, 1], hue='month', palette='viridis', legend=False)
axes[1, 1].set_title('Passenger Distribution by Month (Box Plot)')
axes[1, 1].set_xlabel('Month')
axes[1, 1].set_ylabel('Passengers')
axes[1, 1].tick_params(axis='x', rotation=45) # Rotates month labels so they don't overlap

plt.tight_layout()
plt.show()
```

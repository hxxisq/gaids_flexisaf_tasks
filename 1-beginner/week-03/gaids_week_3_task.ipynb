```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
```

```python
# Load the dataset
iris = load_iris()

# Create the DataFrame with feature columns
data = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add the species column using categorical codes
data['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

```python
data.head()
```

```python
data.describe().round(3)
```

```python
numerical_cols = iris.feature_names

# Mean, Median, Mode & Std Dev per Feature
stats = []
for col in numerical_cols:
    series = data[col]
    stats.append({
        "Feature":    col,
        "Mean":       round(series.mean(), 4),
        "Median":     round(series.median(), 4),
        "Mode":       round(series.mode()[0], 4),
        "Std Dev":    round(series.std(), 4),
    })

stats_df = pd.DataFrame(stats).set_index("Feature")
display(stats_df)
```

```python
# Correlation Matrix
corr_matrix = data[numerical_cols].corr().round(4)
display(corr_matrix)
```

```python
#  Notable Correlations (|r| > 0.8
corr_pairs = (
    corr_matrix
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)
corr_pairs.columns = ["Feature A", "Feature B", "r"]
strong = corr_pairs[corr_pairs["r"].abs() > 0.8].sort_values("r", ascending=False)
if strong.empty:
    print("  No strong correlations found.")
else:
    for _, row in strong.iterrows():
        direction = "positive" if row["r"] > 0 else "negative"
        print(f"  {row['Feature A']}  ↔  {row['Feature B']}")
        print(f"    r = {row['r']:.4f}  ({direction} correlation)\n")
```
